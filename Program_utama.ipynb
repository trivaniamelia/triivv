```python
# ==========================================
# PROGRAM PENGELOLAAN HASIL PANEN
# ==========================================

# Scope: Variable Global
NAMA_TEMPAT = "Gudang Utama Tani Makmur"


def inisialisasi_data():
    """1. Fungsi inisialisasi menggunakan dictionary, list, set, dan tuple."""
    # Dict: Menyimpan data komoditas {id: info}
    data_komoditas = {
        "P01": {
            "nama": "Padi",
            "jumlah": 1500.0,
            "harga": 6000.0,
            "lokasi": ("Silo A", 1),
        },  # Tuple
        "P02": {
            "nama": "Jagung",
            "jumlah": 800.0,
            "harga": 4500.0,
            "lokasi": ("Silo B", 2),
        },
    }

    # Set: Menyimpan jenis komoditas unik
    kategori_panen = {"Biji-bijian", "Sereal"}

    # List: Log riwayat transaksi
    riwayat_transaksi = ["Inisialisasi sistem berhasil."]

    return data_komoditas, kategori_panen, riwayat_transaksi


def tambah_hasil_panen(
    data_gudang,
    kategori_set,
    riwayat_list,
    kode_id,
    nama,
    jumlah,
    harga,
    kategori="Biji-bijian",
    blok="Silo A",
    rak=1,
):
    """2. Fungsi penambahan data dengan Default Argument & Validasi Input."""
    # Validasi input (Edge cases)
    if not kode_id or not nama:
        raise ValueError("ID dan Nama komoditas tidak boleh kosong.")
    if jumlah < 0 or harga < 0:
        raise ValueError("Jumlah dan harga tidak boleh bernilai negatif.")

    # Scope: Variable Lokal
    info_lokasi = (blok, rak)  # Tuple

    # Menambahkan/memperbarui data dictionary
    if kode_id in data_gudang:
        data_gudang[kode_id]["jumlah"] += jumlah
        data_gudang[kode_id]["harga"] = harga
        status = "diperbarui"
    else:
        data_gudang[kode_id] = {
            "nama": nama,
            "jumlah": jumlah,
            "harga": harga,
            "lokasi": info_lokasi,
        }
        status = "ditambahkan"

    kategori_set.add(kategori)
    riwayat_list.append(f"Komoditas {nama} ({kode_id}) berhasil {status}.")
    return True


def kurangi_jumlah_panen(data_gudang, riwayat_list, kode_id, jumlah_keluar):
    """3. Fungsi untuk mengurangi jumlah panen saat terjadi transaksi keluar/penjualan."""
    if kode_id not in data_gudang:
        raise KeyError("Kode komoditas tidak ditemukan.")
    if jumlah_keluar <= 0:
        raise ValueError("Jumlah pengurangan harus lebih besar dari 0.")

    jumlah_sekarang = data_gudang[kode_id]["jumlah"]
    if jumlah_keluar > jumlah_sekarang:
        raise ValueError("Jumlah tidak mencukupi untuk transaksi.")

    data_gudang[kode_id]["jumlah"] -= jumlah_keluar
    nama = data_gudang[kode_id]["nama"]
    riwayat_list.append(f"Jumlah {nama} ({kode_id}) berkurang sebanyak {jumlah_keluar}.")
    return data_gudang[kode_id]["jumlah"]


def hitung_total_aset(data_gudang):
    """4. Fungsi untuk menghitung total akumulasi nilai aset gudang."""
    total_nilai = 0.0
    for info in data_gudang.values():
        total_nilai += info["jumlah"] * info["harga"]
    return total_nilai


def cetak_laporan_panen(data_gudang, kategori_set, riwayat_list):
    """5. Fungsi visualisasi laporan panen."""
    print(f"\n==========================================")
    print(f" LAPORAN HASIL PANEN: {NAMA_TEMPAT}")
    print(f"==========================================")
    print("Daftar Komoditas:")
    for id_k, val in data_gudang.items():
        lokasi_str = f"{val['lokasi'][0]} - Rak {val['lokasi'][1]}"
        print(
            f"- [{id_k}] {val['nama']:<10} | Jumlah: {val['jumlah']:>7.1f} kg | "
            f"Harga/kg: Rp{val['harga']:>8.2f} | Lokasi: {lokasi_str}"
        )

    print("\nKategori Terdaftar (Set):", ", ".join(kategori_set))
    print(f"Total Aset Gudang    : Rp{hitung_total_aset(data_gudang):,.2f}")
    print("\nLog Riwayat Transaksi (List):")
    for log in riwayat_list:
        print(f" - {log}")
    print("==========================================\n")


def main():
    """Fungsi utama penggerak alur program."""
    print("Memulai Sistem Pengelolaan Hasil Panen...")

    # Inisialisasi
    gudang, kat, log = inisialisasi_data()

    # Tambah data baru (Menggunakan default argument untuk kategori dan lokasi)
    tambah_hasil_panen(
        gudang, kat, log, "P03", "Kedelai", 500.0, 9000.0, kategori="Kacang-kacangan"
    )

    # Tambah data dengan parameter default penuh
    tambah_hasil_panen(gudang, kat, log, "P04", "Gandum", 300.0, 7000.0)

    # Kurangi jumlah
    kurangi_jumlah_panen(gudang, log, "P01", 200.0)

    # Cetak laporan akhir
    cetak_laporan_panen(gudang, kat, log)


if __name__ == "__main__":
    main()
```
